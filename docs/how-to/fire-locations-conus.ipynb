```python
# ============================================================
# 4) Check that all requested points are inside the dataset bounds
# ============================================================
lat_min = float(ds['latitude'].min().compute())
lat_max = float(ds['latitude'].max().compute())
lon_min = float(ds['longitude'].min().compute())
lon_max = float(ds['longitude'].max().compute())

inside = points_df['lat'].between(lat_min, lat_max) & points_df['lon'].between(lon_min, lon_max)

if not inside.all():
    bad = points_df.loc[~inside, ['label', 'state', 'lat', 'lon']]
    raise ValueError(
        f'Some requested points fall outside dataset bounds:\n{bad.to_string(index=False)}'
    )
```
